```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data_dir = "C:/Users/Admin/Desktop/22-Machine_Learning-Projects/data"
df = pd.read_csv(f'{data_dir}/3_BuildLinearRegression(no scikit-learn, just math)/SOCR-HeightWeight.csv')
print(df.shape)
df.head()
```

```python
# Create a scatterplot
sns.scatterplot(
    data = df,
    x = "Height(Inches)",
    y = "Weight(Pounds)"
)
plt.show()
```

```python
# Get x and y
x = df["Height(Inches)"].values
y = df["Weight(Pounds)"].values
```

```python
# Find m and b
N = x.shape[0]
m = (N * np.sum(x * y) - np.sum(x) * np.sum(y)) / (N * np.sum(x**2) - (np.sum(x) ** 2))
b = (np.sum(y) - m * np.sum(x)) / N
print(m, b)
```

```python
# Draw regression line
x_min = np.min(x)
y_min = m * x_min + b
x_max = np.max(x)
y_max = m * x_max + b

fig, ax = plt.subplots()
sns.scatterplot(
    data = df,
    x = "Height(Inches)",
    y= "Weight(Pounds)",
    ax = ax,
    alpha = 0.4
)

sns.lineplot(
    x = [x_min, x_max],
    y = [y_min, y_max],
    linewidth = 1.5,
    color ='red'
)
plt.show()
```
